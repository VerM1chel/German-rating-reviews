```python
# Убираем ссылки
train.Text = train.Text.str.replace(r"https?:\/\/\S+|www\.\S+", "", regex=True)
# Убираем теги
train.Text = train.Text.str.replace(r"[@#]\w+", "", regex=True)
# Убираем повторы букв (например, "langsaaaaaam" в "langsam")
train.Text = train.Text.str.replace(r"(\w)\1+", r"\1", regex=True)
```
